#### Tokenisation avec des expressions régulières


Ce code montre comment diviser une chaîne de texte en conservant les délimiteurs (espaces) grâce aux expressions régulières.

```python
import re
text = "Hello, world. This, is a test."
result = re.split(r'(\s)', text)
print(result)
```

Modification de l'expression régulière pour diviser selon les espaces (\s), les virgules et les points.
Cette regex `([,.]|\s)` capture soit une ponctuation (virgule ou point) soit un espace

```python
import re
text = "Hello, world. This, is a test."
result = re.split(r'([,.]|\s)', text)
print(result)
```

Suppression des caractères d'espacement redondants.
On ne garde que les éléments qui ne sont pas vides après avoir retiré les espaces :

```python
import re
text = "Hello, world. This, is a test."
result = re.split(r'([,.]|\s)', text)
result = [item for item in result if item.strip()]
print(result)
```

Tokenisation avec plusieurs types de caractères spéciaux : 


```python
text = "Hello, world. Is this-- a test?"
result = re.split(r'([,.:;?_!"()\']|--|\s)', text)
result = [item.strip() for item in result if item.strip()]
print(result)
```

### 2.3. Conversion des tokens en tokens IDs


```python
# Création du vocabulaire et affichage de 51 premiers tokens
all_words = sorted(set(preprocessed))
vocab_size = len(all_words)
print(vocab_size)

vocab = {token:integer for integer,token in enumerate(all_words)}
for i, item in enumerate(vocab.items()):
    print(item)
    if i >= 50:
        break
```

#### Implémentation complète d'un tokenizer simple


```python
class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}
    def encode(self, text):
        preprocessed = re.split(r'([,.?_!"()\']|--|\s)', text)
        preprocessed = [
        item.strip() for item in preprocessed if item.strip()
        ]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
```

```python
# Utilisation du tokenizer
tokenizer = SimpleTokenizerV1(vocab)
text = """"It's the last he painted, you know,"
Mrs. Gisburn said with pardonable pride."""
ids = tokenizer.encode(text)
print(ids)
```

```python
# Décodage des IDs en texte
decoded_text = tokenizer.decode(ids)
print(decoded_text)
```

Ce tokenizer est simple mais il ne gère pas les tokens manquants dans le dataset d'entraînement.


```python
text = "Hello, do you like tea?"
try:
    print(tokenizer.encode(text))
except KeyError as e:
    print(f"Token manquant: {e}")
```

### 2.4. Ajout des tokens spéciaux


Nous pouvons modifier le tokenizer pour utiliser un token <|unk|> (inconnu) lorsqu'il rencontre un mot qui ne fait pas partie du vocabulaire. De plus, nous ajoutons un token pour séparer les textes non liés.

```python
all_tokens = sorted(list(set(preprocessed)))
all_tokens.extend(["<|endoftext|>", "<|unk|>"])
vocab = {token:integer for integer,token in enumerate(all_tokens)}
print(len(vocab.items()))
```

```python
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = { i:s for s,i in vocab.items()}
    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [
        item.strip() for item in preprocessed if item.strip()
        ]
        preprocessed = [item if item in self.str_to_int
        else "<|unk|>" for item in preprocessed]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text
```

```python
# Utilisation du tokenizer
tokenizer = SimpleTokenizerV2(vocab)
text1 = """"It's the last he painted, you know,"
Mrs. Gisburn said with pardonable pride."""
text2 = "Hello, do you like tea?"
text = text1 + " <|endoftext|> " + text2
ids = tokenizer.encode(text)
print(ids)

# Décodage des IDs en texte
decoded_text = tokenizer.decode(ids)
print(decoded_text)
```
