#### Tokenisation avec des expressions régulières


Ce code montre comment diviser une chaîne de texte en conservant les délimiteurs (espaces) grâce aux expressions régulières.

```python
import re
text = "Hello, world. This, is a test."
result = re.split(r'(\s)', text)
print(result)
```

Modification de l'expression régulière pour diviser selon les espaces (\s), les virgules et les points.
Cette regex `([,.]|\s)` capture soit une ponctuation (virgule ou point) soit un espace

```python
import re
text = "Hello, world. This, is a test."
result = re.split(r'([,.]|\s)', text)
print(result)
```

Suppression des caractères d'espacement redondants.
On ne garde que les éléments qui ne sont pas vides après avoir retiré les espaces :

```python
import re
text = "Hello, world. This, is a test."
result = re.split(r'([,.]|\s)', text)
result = [item for item in result if item.strip()]
print(result)
```

```python
text = "Hello, world. Is this-- a test?"
result = re.split(r'([,.:;?_!"()\']|--|\s)', text)
result = [item.strip() for item in result if item.strip()]
print(result)
```

### 2.3. Conversion des tokens en tokens IDs


```python
# Création du vocabulaire et affichage de 51 premiers tokens
all_words = sorted(set(preprocessed))
vocab_size = len(all_words)
print(vocab_size)

vocab = {token:integer for integer,token in enumerate(all_words)}
for i, item in enumerate(vocab.items()):
    print(item)
    if i >= 50:
        break
```

#### Implémentation complète d'un tokenizer simple


```python
class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}
    def encode(self, text):
        preprocessed = re.split(r'([,.?_!"()\']|--|\s)', text)
        preprocessed = [
        item.strip() for item in preprocessed if item.strip()
        ]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
```

```python
# Utilisation du tokenizer
tokenizer = SimpleTokenizerV1(vocab)
text = """M. Myriel était arrivé à Digne accompagné d’une vieille fille, 
mademoiselle Baptistine, qui était sa sœur et qui avait dix ans de moins que lui."""
ids = tokenizer.encode(text)
print(ids)
```

```python
# Décodage des IDs en texte
decoded_text = tokenizer.decode(ids)
print(decoded_text)
```

### 2.4. Ajout des tokens spéciaux


Nous pouvons modifier le tokenizer pour utiliser un token <|unk|> (inconnu) lorsqu'il rencontre un mot qui ne fait pas partie du vocabulaire. De plus, nous ajoutons un token pour séparer les textes non liés.

```python
all_tokens = sorted(list(set(preprocessed)))
all_tokens.extend(["<|endoftext|>", "<|unk|>"])
vocab = {token:integer for integer,token in enumerate(all_tokens)}
print(len(vocab.items()))
```

```python
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = { i:s for s,i in vocab.items()}
    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [
        item.strip() for item in preprocessed if item.strip()
        ]
        preprocessed = [item if item in self.str_to_int
        else "<|unk|>" for item in preprocessed]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text
```

```python
# Utilisation du tokenizer
tokenizer = SimpleTokenizerV2(vocab)
text1 = """M. Myriel était arrivé à Digne accompagné d’une vieille fille, 
mademoiselle Baptistine, qui était sa sœur et qui avait dix ans de moins que lui."""
text2 = "Hello, do you like tea?"
text = text1 + " <|endoftext|> " + text2
ids = tokenizer.encode(text)
print(ids)

# Décodage des IDs en texte
decoded_text = tokenizer.decode(ids)
print(decoded_text)
```

```python
context_size = 4
x = enc_sample[:context_size]
y = enc_sample[1:context_size+1]
print(f"x: {x}")
print(f"y: {y}")
```

Pour l'implémentation efficace du chargeur de données, nous utiliserons les 
classes Dataset et DataLoader intégrées à PyTorch. 

```python
### !pip install torch
```

```python
# A dataset for batched inputs and targets
import torch
from torch.utils.data import Dataset, DataLoader
class GPTDatasetV1(Dataset):
    def __init__(self, txt, tokenizer, max_length, stride):
        self.input_ids = []
        self.target_ids = []
        token_ids = tokenizer.encode(txt)

        for i in range(0, len(token_ids) - max_length, stride):
            input_chunk = token_ids[i:i + max_length]
            target_chunk = token_ids[i + 1: i + max_length + 1]
            self.input_ids.append(torch.tensor(input_chunk))
            self.target_ids.append(torch.tensor(target_chunk))
    def __len__(self):
        return len(self.input_ids)
    def __getitem__(self, idx):
        return self.input_ids[idx], self.target_ids[idx]
```

```python
# A data loader to generate batches with input-with pairs

def create_dataloader_v1(txt, batch_size=4, max_length=256,
stride=128, shuffle=True, drop_last=True,
num_workers=0):
    tokenizer = tiktoken.get_encoding("gpt2")
    dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)
    dataloader = DataLoader(
    dataset,
    batch_size=batch_size,
    shuffle=shuffle,
    drop_last=drop_last,
    num_workers=num_workers
    )
    return dataloader
```

```python
with open("les_miserables.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()
dataloader = create_dataloader_v1(
raw_text, batch_size=1, max_length=4, stride=1, shuffle=False)
data_iter = iter(dataloader)
first_batch = next(data_iter)
print(first_batch)
```
